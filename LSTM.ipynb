## Task 1: Load and Understand the Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

try:
    df = pd.read_csv('movie_reviews.csv')
    print("Dataset loaded successfully!")
except FileNotFoundError:
    print("Error: 'movie_reviews.csv' not found. Please upload the file or update the path.")
    data = {'review': ['The movie was amazing and beautifully directed.', 'I really enjoyed the story and acting.', 'The movie was boring and too slow.', 'The acting was terrible and the story was weak.', 'This film is a masterpiece of storytelling.', 'Absolutely dreadful and a waste of time.', 'A truly unforgettable cinematic experience.', 'Could not finish watching, it was so bad.', 'Highly recommended for its unique plot.', 'The plot was convoluted and performances were flat.'],
            'sentiment': ['Positive', 'Positive', 'Negative', 'Negative', 'Positive', 'Negative', 'Positive', 'Negative', 'Positive', 'Negative']}
    df = pd.DataFrame(data)
    print("Created a dummy DataFrame for demonstration.")
```

```python
display(df.head(10))
```

```python
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
sentiment_distribution = df['sentiment'].value_counts()
print("Sentiment Distribution:")
print(sentiment_distribution)

plt.figure(figsize=(6, 4))
sns.barplot(x=sentiment_distribution.index, y=sentiment_distribution.values, palette='viridis')
plt.title('Distribution of Positive and Negative Reviews')
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')
plt.show()
```

```python
print(f"Number of unique sentiment classes: {df['sentiment'].nunique()}")
```
